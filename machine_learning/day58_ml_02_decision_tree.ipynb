## 의사결정나무
- 질문에 따라 판단
- 젤 위 루트 노드 --> 분할 --> 분할 ...
- 깊이가 깊어질때 과적합이 발생할 위험이 크다.
- 분류/회귀 둘 다 가능하다.

```python
# 실습 폴더 불러오기
from pathlib import Path
import warnings
warnings.filterwarnings('ignore')

DATA_DIR = Path('ml_data')
if not DATA_DIR.exists():
    DATA_DIR = path('ml_data')
print(f'데이터 폴더: {DATA_DIR.resolve()}')
```

### 라이브러리 불러오기

```python
import pandas as pd
from sklearn.model_selection import train_test_split
# export_text : 나무 구조를 글자로 출력
from sklearn.tree import DecisionTreeClassifier, export_text # 의사결정나무 분류
from sklearn.impute import SimpleImputer  # 결측치를 특정 규칙에 따라 채워주는 전처리 도구
from sklearn.pipeline import make_pipeline # 여러 전처리 단계와 모델을 하나의 객체로 순서대로 연결
```

### 데이터 불러오기

```python
df=pd.read_csv(DATA_DIR/'telecom_churn.csv')
df.head()
```

### 피처(입력) / 타겟(정답) 데이터 나누기

```python
# usage_minutes : 고객의 (월간) 통화/이용 시간(분) -> 이용량이 급격히 줄어든 고객은 이미 다른 서비스로 떠났다.
# complaints : 고객이 접수한 불만/민원 건수 --> 서비스에 불만족해서 이탈할 확률이 높다고 판단
# contract_months: 계약 유지 기간(개월 수) --> 계약 기간이 짧을 수록 이탈 위험이 큰 것으로 판단
# monthly_fee : 월 요금 (청구액) --> 요금이 부담스러울수록 이탈 가능성이 크다.
# churn : 타겟(정답), 이탈한다(1)/안한다(0)
features = ['usage_minutes', 'complaints', 'contract_months', 'monthly_fee']
X = df[features]
y = df['churn']

X.shape, y.shape
```

### 훈련 / 검증 데이터 나누기

```python
X_train, X_vaild, y_train, y_valid = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

X_train.shape, X_vaild.shape, y_train.shape, y_valid.shape
```

### 의사결정나무 분류 모델 학습시키기

```python
model = DecisionTreeClassifier(random_state=42)  # 일관성있는 테스트 데이터 추출
model.fit(X_train, y_train)  # 학습하기
```

```python
pred = model.predict(X_vaild) # 예측값
pred
```

```python
from sklearn.metrics import accuracy_score # 정확도

accuracy_score(y_valid, pred) # 정답과 예측을 비교해서 얼마나 정확한가?
```

### 나무 깊이를 바꿔가며 훈련/검증 점수를 비교 --> 과적합을 관찰하기 위해서..

```python
# 깊이를 2, 4, 제한없음(None) 세 가지로 바꿔가며 성능 비교
for depth in [2, 4, None]:
    model = make_pipeline(
        SimpleImputer(strategy='median'), # 결측치 패턴을 중앙값으로 채운다.
        DecisionTreeClassifier(max_depth=depth, random_state=42)
    )
    model.fit(X_train, y_train)  #학습
    pred = model.predict(X_vaild) #예측
    print(f'{depth} -> {accuracy_score(y_valid, pred):.4f}') #정확도 확인(소수 네째자리까지)
```

- 깊이가 없음(끝까지 내려가는 것)이 깊이 2와 성능이 서로 비슷하다면 깊이가 얕은 모델을 우선한다.

### 추가) 깊이가 2인 얖은 나무모델을 직접 뜯어보기

```python
shallow = make_pipeline(        # 여러 작업을 순서대로 실행하는 하나의 파이프라인을 만드는 함수
    SimpleImputer(strategy='median'),  # 결측치(NaN)가 있으면 중앙값(median) 으로 채워줍니다.
    DecisionTreeClassifier(max_depth=2, random_state=42)
) 
shallow.fit(X_train, y_train)  # 파이프라인 전체를 학습
# named_steps : 파이프라인 안에서 이름으로 특정 단계(모델)를 꺼내오는 방법
tree_model = shallow.named_steps['decisiontreeclassifier']

print(export_text(tree_model, feature_names=features))
```

## 여러 모델의 의견을 모으는 앙상블
- 랜덤포레스트
    - 데이터를 무작위로 복원추출(부트스트랩), 
    - 각 분할마다 특성도 무작위로 일부만 사용해서 서로 다른 여러 개를 만든 뒤 평균/다수결로 예측
    - 단일 나무보다 과적합에 강하지만, 개별 나무 하나하나를 사람이 읽고 설명하기는 어려워진다라는 트레이드모프가 있다.
    - "여러 사람에게 물어보고 다수결로 정한다"라는 비유
    - 분류/회귀 둘 다 사용 가능
    - 분산 처리(배깅), 다수결(보딩)

### 라이브러리 불러오기

```python
from sklearn.ensemble import RandomForestClassifier  # 여러 개의 의사결정나무를 모아 만드는 앙상블 모델
from sklearn.metrics import f1_score # 정밀도와 재현율의 조화평균(불균형 데이터로 적합하다.)
```

```python
df = pd.read_csv(DATA_DIR/'telecom_churn.csv')
features
```

```python
# 입력(피처), 정답(타겟) --> 훈련/검증 나누기
X = df[features]
y = df['churn']
X_train, X_vaild, y_train, y_valid = train_test_split(
    X, y, test_size=0.25, stratify=y, random_state=42
)
X_train.shape, X_vaild.shape, y_train.shape, y_valid.shape
```

```python
model = RandomForestClassifier(
    n_estimators=200,    # 나무가 200그루
    max_depth=5,
    random_state=42
)
# 학습
model.fit(X_train, y_train)
```

```python
# 예측
pred = model.predict(X_vaild)

# 평가 - 정확도
print(f'랜덤포레스트 분류의 정확도: {accuracy_score(y_valid, pred):.4f}')
```

### 의사결정나무와 랜덤포레스트 비교

```python

models = {
    'tree' : DecisionTreeClassifier(max_depth=4, random_state=42),
    'forest' : RandomForestClassifier(n_estimators= 200, max_depth=4 , random_state=42)
}
models.keys()
```

```python
models.values()
```

```python

models.items() # 튜플로  쌍을 묵는다.
```

```python
fitted={}
for name, estimator in models.items():
    pipe = make_pipeline(SimpleImputer(strategy='median'),estimator)
    pipe.fit(X_train, y_train) # 학습
    fitted[name] =pipe  # 나중에 다시 꺼내쓰기 위해서 딕셔너리에 저장
    pred = pipe.predict(X_vaild) # 예측
    print(f'{name} F1 score: {f1_score(y_valid,pred):.4f}')  # f1 score(조화평균) 확인
    print(f'{name} 정확도: {accuracy_score(y_valid, pred):.4f}') # 정확도 확인
```

```python
fitted
```

### 랜덤포레스트가 어떤 특성을 중요하게 봤는지 확인

```python
forest = fitted['forest'].named_steps['randomforestclassifier']
forest
```

```python
importance = forest.feature_importances_   # 맨뒤에 - 가 붙는것은 학습한 결과, 특성 중요도
importance
```
