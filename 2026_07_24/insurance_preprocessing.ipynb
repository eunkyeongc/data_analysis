### 결측치 처리

```python
# 결측치 확인하기
df.isnull().sum()
```

```python
# 결측 비율이 50% 초과 --> 컬럼을 삭제
# 결측치 비율을 알기 위해 평균 구하기
missing_ratio = df.isnull().mean()
missing_ratio
```

```python
missing_ratio[missing_ratio > 0.5]
```

```python
missing_ratio[missing_ratio > 0.02]
```

```python
type(missing_ratio[missing_ratio > 0.02])
```

```python
type(df)
```

```python
drop_cols = missing_ratio[missing_ratio > 0.5].index.tolist()
```

```python
df.drop(columns=drop_cols)  # df 데이터프레임에서 drop_cols 리스트에 들어 있는 컬럼들을 삭제한 새로운 데이터프레임을 반환
```

```python
df.select_dtypes(include='number').columns #  데이터 타입이 'number'(숫자형)인 컬럼만 보여줘.
```
