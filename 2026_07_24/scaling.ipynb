### 스케일링(scaling)
- 컬럼 간 숫자의 상대적 크기 차이 때문에, 모델에 영향을 미치는 문제들이 달라진다. 이를 해결하기 위한 과정이다.
- 정규화(Normalization), 표준화(Standardization) 이 두가지를 가장 많이 활용한다.
    - 정규화(Normalization) : 데이터를 일정한 범위(주로 0~1)로 변환, 대표적인 방법은 Min-Max Scaling
    - 표준화(Standardization) : 평균 0, 표준편차 1이 되도록 변환, 대표적인 방법은 Z-score Standardization
- 사이킷런 라이브러리에 구현되어 있다.

### 라이브러리 불러오기

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler, StandardScaler # 정규화 표준화
```

### 데이터 불러오기

```python
CSV_PATH = 'Clean_Dataset.csv'
df_origin = pd.read_csv(CSV_PATH)

df_origin.head()
```

```python
df = df_origin.drop('Unnamed: 0', axis=1)
df.info()
```

```python
# 수치형 데이터만 분리하여 데이터프레임 만들기
df_num = df[['duration', 'days_left', 'price']]
df_num
```

### 정규화 수식 적용하기 - 수식만으로 해보기

```python
df_rum_norm = (df_num - df_num.min()) / (df_num.max() - df_num.min())
df_rum_norm.head()      # 모든 수치데이터가 0~1 사이로 정규화됨
```

### 사이킷런으로 동일 결과 재현하기 --MinMaxScaler

```python
minmax = MinMaxScaler()
minmax.fit_transform(df_num)
```

```python
result1 = minmax.fit_transform(df_num)

# 데이터프레임으로 만들어서 결과확인하기
df_num_norm_sklearn = pd.DataFrame(result1, columns=df_num.columns)
df_num_norm_sklearn.head()
```

### 표준화 수식 적용하기

```python
df_num
```

### 1. 표준화 수식 적용하기
- 정확히 0이 아니라 아주 작은 값으로 나오는 이유는??
    - 부동소수점 연산 오차 문제
    - 정규화와 다르게 0과 1로 고정되어 있지 않고, 데이터 분포에 따라 달라진다는 특징도 있다.

- 이상치가 많거나 데이터 범위가 정해지지 않았다면 표준화가 더 안전한 편.
- 데이터 범위가 명확히 정해져있고, 이상치가 없다면 정규화도 무방하다.

```python
df_num_std =(df_num - df_num.mean()) / df_num.std()   # std() - 표준편차, mean() - 평균
df_num_std
```

### 2. 사이킷런으로 표준화해보기 - StandardScaler

```python
result2 = StandardScaler().fit_transform(df_num)
result2
```

```python
std_scaler = StandardScaler()
result2 = std_scaler.fit_transform(df_num)
result2
```

```python
# 데이터프레임으로 다시한번 결과 확인하기
df_num_std_sklearn = pd.DataFrame(
    result2,
    columns=df_num.columns
)
df_num_std_sklearn.head()
```
